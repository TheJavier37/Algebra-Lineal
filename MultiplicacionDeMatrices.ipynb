```python
import numpy as np

# Definir dos matrices A y B de 3x3
A = np.array([[1, 0],
              [6, -2],
              [-1, 4]])

B = np.array([[2, -5, 1, -2],
              [3, 2, 0, -1]])

print("Matriz A:\n", A)
print("\nMatriz B:\n", B)

# ------------------------------
# Suma de matrices
# ------------------------------
# suma = A + B
# print("\n Suma (A + B):\n", suma)

# ------------------------------
#  Resta de matrices
# ------------------------------
# resta = A - B
# print("\n Resta (A - B):\n", resta)

# ------------------------------
# Multiplicación por un escalar
# ------------------------------
# escalar = 2
# producto_escalar = escalar * A
# print(f"\n Multiplicación por escalar ({escalar} * A):\n", producto_escalar)

# ------------------------------
#  Multiplicación matricial
# ------------------------------
producto = np.dot(A, B)   # o A @ B
print("\n Producto matricial (A * B):\n", producto)

valor = A[2,0]*B[0,3]+A[2,1]*B[1,3]
print("El valor de C[2,3] es igual a: ", valor)

# ------------------------------
# Matriz transpuesta
# ------------------------------
# transpuesta = A.T
# rint("\n Transpuesta de A (Aᵀ):\n", transpuesta)
```
