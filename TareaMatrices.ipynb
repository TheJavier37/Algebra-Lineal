```python
import numpy as np

# Definir dos matrices A y B de 3x3
A = np.array([[2, 4, 6],
              [1, 3, 5],
              [0, -1, 2]])

B = np.array([[3, 2, 1],
              [4, 0, -2],
              [1, 3, 5]])

print("Matriz A:\n", A)
print("\nMatriz B:\n", B)

# ------------------------------
# Suma de matrices
# ------------------------------
suma = A + B
print("\n Suma (A + B):\n", suma)
print("\n C[0,0] es igual a: ", A[0,0]+B[0,0])

# ------------------------------
#  Resta de matrices
# ------------------------------
resta = A - B
print("\n Resta (A - B):\n", resta)
print("\n C[2,1] es igual a: ", A[2,1]-B[2,1])

# ------------------------------
# Multiplicación por un escalar
# ------------------------------
escalar = 2
producto_escalar = escalar * A
print(f"\n Multiplicación por escalar ({escalar} * A):\n", producto_escalar)
print("\n A[2,1] es igual a: ", 2 * A[2,1])

# ------------------------------
#  Multiplicación matricial
# ------------------------------
producto = np.dot(A, B)   # o A @ B
print("\n Producto matricial (A * B):\n", producto)
print("\n C[2,2] es igual a: ", A[2,0]*B[0,2]+A[2,1]*B[1,2]+A[2,2]*B[2,2])

# ------------------------------
# Matriz transpuesta
# ------------------------------
# transpuesta = A.T
# print("\n Transpuesta de A (Aᵀ):\n", transpuesta)
```
