```python
from sympy import symbols, Eq, solve
x, y, z = symbols('x y z')
sol = solve([
    Eq(x + y + z, 1200),
    Eq(2*x + y + 3*z, 2500),
    Eq(x - y + z, 400)
], (x, y, z))
sol
```
