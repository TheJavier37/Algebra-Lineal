```python
from sympy import symbols, Eq, solve
x, y, z = symbols('x y z')
sol = solve([
    Eq(x + y + z, 1200),
    Eq(2*x + y + 3*z, 2500),
    Eq(x - y + z, 400)
], (x, y, z))
sol
```

```python
from sympy import symbols, Eq, solve
x, y, z = symbols('x y z')
sol = solve([
    Eq(x + 2*y + 3*z, 14),
    Eq(2*x + 3*y + z, 13),
    Eq(3*x + y + 2*z, 13)
], (x, y, z))
sol
```
