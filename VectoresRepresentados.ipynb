```python
import datetime

nombre = "Javier Guarnizo"
fecha_actual = datetime.date.today()
carrera = "Ingeniería de Ciencias de la Computacion"

texto_salida = f"Nombre: {nombre}\nFecha: {fecha_actual.strftime('%d/%m/%Y')}\nCarrera: {carrera}"

print(texto_salida)

import matplotlib.pyplot as plt
import numpy as np

# Origen
origin = np.array([0, 0])

# Crear la figura y los ejes para un gráfico amplio
fig, ax = plt.subplots(figsize=(12, 12))

# Dibujar los vectores originales
ax.quiver(*origin, *vector1, angles='xy', scale_units='xy', scale=1, color='blue', label='Vector 1 (original)', alpha=0.7)
ax.quiver(*origin, *vector2, angles='xy', scale_units='xy', scale=1, color='green', label='Vector 2 (original)', alpha=0.7)

# Dibujar la suma de vectores (que es la resta actual)
ax.quiver(*origin, *suma_vector, angles='xy', scale_units='xy', scale=1, color='red', linewidth=2, label='Resta de Vectores (vector1 - vector2)')

# Dibujar el primer vector escalado
ax.quiver(*origin, *scaled_vector, angles='xy', scale_units='xy', scale=1, color='purple', linewidth=2, label='Vector Escaldo (0.5 * vector1)')

# Dibujar el segundo vector escalado
ax.quiver(*origin, *scaled_vector2, angles='xy', scale_units='xy', scale=1, color='orange', linewidth=2, label='Vector Escaldo (-4 * vector2)')

# Configurar el gráfico
ax.set_xlabel('Eje X')
ax.set_ylabel('Eje Y')
ax.set_title('Representación Gráfica de Vectores Originales, Suma y Escalados')
ax.grid()
ax.axhline(0, color='black', linewidth=0.5)
ax.axvline(0, color='black', linewidth=0.5)

# Calcular límites máximos para todos los vectores para asegurar que todos sean visibles
all_vectors_data = np.array([vector1, vector2, suma_vector, scaled_vector, scaled_vector2])
max_abs_coord = np.max(np.abs(all_vectors_data)) * 1.3 # Un factor para dar más espacio alrededor

ax.set_xlim([-max_abs_coord, max_abs_coord])
ax.set_ylim([-max_abs_coord, max_abs_coord])
ax.set_aspect('equal', adjustable='box')
ax.legend()

# Añadir los datos del usuario al gráfico
ax.text(0.02, 0.98, f"Nombre: {nombre}", transform=ax.transAxes, verticalalignment='top', fontsize=10)
ax.text(0.02, 0.95, f"Fecha: {fecha_actual.strftime('%d/%m/%Y')}", transform=ax.transAxes, verticalalignment='top', fontsize=10)
ax.text(0.02, 0.92, f"Carrera: {carrera}", transform=ax.transAxes, verticalalignment='top', fontsize=10)

plt.show()
```
